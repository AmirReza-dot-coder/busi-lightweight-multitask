# خروجی گرفتن از وزن‌های مدل برای بررسی استاد
این نوت‌بوک فقط فایل‌های ثبت‌شده را کپی و SHA256 آن‌ها را کنترل می‌کند؛ آموزش یا inference انجام نمی‌دهد. قبل از اجرا فضای کافی در Drive داشته باش. فایل‌ها به‌صورت محلی در Drive خودت ذخیره می‌شوند و اشتراک‌گذاری خودکار انجام نمی‌شود.


```python
from google.colab import drive
drive.mount('/content/drive')
```

از بستهٔ پروژه، دو فایل `tools/export_checkpoints.py` و `data/required_checkpoints.json` را انتخاب کن.


```python
from google.colab import files
from pathlib import Path
uploaded = files.upload()
for name in ['export_checkpoints.py', 'required_checkpoints.json']:
    if name not in uploaded:
        raise ValueError('این فایل را نیز انتخاب کن: ' + name)
    (Path('/content') / name).write_bytes(uploaded[name])
```

```python
import subprocess, sys
subprocess.run([sys.executable, '/content/export_checkpoints.py',
    '--inventory', '/content/required_checkpoints.json',
    '--experiments-root', '/content/drive/MyDrive/BUSI_EXPERIMENTS',
    '--destination', '/content/drive/MyDrive/BUSI_ADVISOR_CHECKPOINTS'], check=True)
```

اگر همهٔ موارد MATCH شدند، پوشهٔ `BUSI_ADVISOR_CHECKPOINTS` و فایل `exported_manifest.json` آمادهٔ بررسی‌اند. در صورت نیاز لینک پوشه را به استاد بده؛ دسترسی آن را خودت در Drive تعیین کن. این خروجی شامل ۲۸ رکورد ثبت‌شده و در حالت کامل ۲۷ فایل وزن یکتاست.
